```python
import warnings
warnings.filterwarnings("ignore", "Wswiglal-redir-stdio")

import os

import numpy as np
# The first import of matplotlib can take some time (especially on cloud platforms). This is normal.
import matplotlib
import matplotlib.pyplot as plt

from gwpy.timeseries import TimeSeries

file_name = "challenge1.gwf"
URL = "https://dcc.ligo.org/public/0187/G2300818/001/" + file_name

channel_name = "H1:CHALLENGE1"
```

```python
if not os.path.isfile(file_name):
  print('Getting : {}'.format(URL))
  import urllib
  urllib.request.urlretrieve(URL, file_name)
else:
  print("No downloading")
```

```python
# Read and plot data
ts = TimeSeries.read(file_name, channel_name)
# Answer to questions 1 and 2
print(f"Signal: duration={ts.duration}, delta_t={ts.dt}, (sampling rate={ts.sample_rate})")
ts.plot();
```

```python
# Response to question 3
hq = ts.q_transform(frange=(30, 500))
plot = hq.plot()
ax = plot.gca()
ax.set_xlim(60, 80)
ax.set_ylim(30, 300)
ax.set_yscale("log")
plot.colorbar(label="Normalised energy");
```
